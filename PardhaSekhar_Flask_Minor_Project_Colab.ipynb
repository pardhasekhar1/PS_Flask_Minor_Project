## Project idea
A Student Registration System. Users enter a name, email, age and course. The application validates the information and displays a registration result. Student records are stored temporarily in memory while the app runs.

```python
import os
os.makedirs('Flask_Minor_Project/templates', exist_ok=True)
open('Flask_Minor_Project/app.py','w',encoding='utf-8').write('from flask import Flask, render_template, request\n\napp = Flask(__name__)\nstudents = []  # Temporary in-memory storage\n\n\n@app.route("/")\ndef home():\n    """GET: display the registration form."""\n    return render_template("index.html")\n\n\n@app.route("/about")\ndef about():\n    """GET: display project information."""\n    return render_template("about.html")\n\n\n@app.route("/students")\ndef view_students():\n    """GET: display students registered during this session."""\n    return render_template("students.html", students=students)\n\n\n@app.route("/register", methods=["POST"])\ndef register():\n    """POST: validate and process the registration form."""\n    name = request.form.get("name", "").strip()\n    email = request.form.get("email", "").strip()\n    course = request.form.get("course", "").strip()\n    age_text = request.form.get("age", "").strip()\n\n    if not name or not email or not course or not age_text:\n        return render_template("index.html", error="Please fill in every field.")\n\n    if "@" not in email or "." not in email.split("@")[-1]:\n        return render_template("index.html", error="Please enter a valid email address.")\n\n    try:\n        age = int(age_text)\n    except ValueError:\n        return render_template("index.html", error="Age must be a whole number.")\n\n    if age < 16 or age > 100:\n        return render_template("index.html", error="Age must be between 16 and 100.")\n\n    student = {"name": name, "email": email, "course": course, "age": age}\n    students.append(student)\n    return render_template("result.html", student=student)\n\n\nif __name__ == "__main__":\n    app.run(host="0.0.0.0", port=5000, debug=False)\n')
open('Flask_Minor_Project/templates/index.html','w',encoding='utf-8').write('<!DOCTYPE html>\n<html lang="en">\n<head>\n<meta charset="UTF-8">\n<meta name="viewport" content="width=device-width, initial-scale=1.0">\n<title>Student Registration</title>\n<style>\nbody { font-family: Arial, sans-serif; background: #f2f5fa; margin: 0; padding: 30px; }\n.container { max-width: 520px; margin: auto; background: white; padding: 28px; border-radius: 12px; box-shadow: 0 2px 12px #0001; }\nh1 { color: #203864; }\nlabel { display: block; margin-top: 14px; font-weight: bold; }\ninput, select { width: 100%; padding: 10px; margin-top: 6px; box-sizing: border-box; border: 1px solid #bbb; border-radius: 6px; }\nbutton { margin-top: 20px; padding: 11px 18px; border: 0; border-radius: 6px; background: #203864; color: white; cursor: pointer; }\n.error { color: #b00020; background: #ffe9ec; padding: 10px; border-radius: 6px; }\nnav a { margin-right: 14px; }\n</style>\n</head>\n<body>\n<div class="container">\n<h1>Student Registration System</h1>\n<p>Enter your details to register for a course.</p>\n<nav><a href="/">Home</a><a href="/about">About</a><a href="/students">View Students</a></nav>\n{% if error %}<p class="error">{{ error }}</p>{% endif %}\n<form action="/register" method="POST">\n<label for="name">Student Name</label>\n<input type="text" id="name" name="name" placeholder="Enter your name">\n<label for="email">Email</label>\n<input type="text" id="email" name="email" placeholder="Enter your email">\n<label for="age">Age</label>\n<input type="number" id="age" name="age" placeholder="Enter your age">\n<label for="course">Course</label>\n<select id="course" name="course">\n<option value="">Select a course</option>\n<option value="Computer Science">Computer Science</option>\n<option value="Artificial Intelligence">Artificial Intelligence</option>\n<option value="Data Science">Data Science</option>\n<option value="Electronics">Electronics</option>\n</select>\n<button type="submit">Register Student</button>\n</form>\n</div>\n</body>\n</html>\n')
open('Flask_Minor_Project/templates/about.html','w',encoding='utf-8').write('<!DOCTYPE html>\n<html><head><meta charset="UTF-8"><title>About</title></head>\n<body style="font-family:Arial;max-width:700px;margin:40px auto;padding:20px;">\n<h1>About This Project</h1>\n<p>This Student Registration System is a beginner Flask mini project.</p>\n<p>It uses GET routes to display pages and a POST route to process student registration details.</p>\n<p>Basic validation checks required fields, email format and age.</p>\n<a href="/">Back to Home</a> | <a href="/students">View Students</a>\n</body></html>\n')
open('Flask_Minor_Project/templates/result.html','w',encoding='utf-8').write('<!DOCTYPE html>\n<html><head><meta charset="UTF-8"><title>Registration Successful</title></head>\n<body style="font-family:Arial;max-width:700px;margin:40px auto;padding:20px;">\n<h1>Registration Successful!</h1>\n<p>The student has been registered.</p>\n<ul>\n<li><b>Name:</b> {{ student.name }}</li>\n<li><b>Email:</b> {{ student.email }}</li>\n<li><b>Age:</b> {{ student.age }}</li>\n<li><b>Course:</b> {{ student.course }}</li>\n</ul>\n<a href="/">Register Another Student</a> | <a href="/students">View Students</a>\n</body></html>\n')
open('Flask_Minor_Project/templates/students.html','w',encoding='utf-8').write('<!DOCTYPE html>\n<html><head><meta charset="UTF-8"><title>Registered Students</title></head>\n<body style="font-family:Arial;max-width:800px;margin:40px auto;padding:20px;">\n<h1>Registered Students</h1>\n{% if students %}\n<table border="1" cellpadding="8" cellspacing="0">\n<tr><th>Name</th><th>Email</th><th>Age</th><th>Course</th></tr>\n{% for student in students %}\n<tr><td>{{ student.name }}</td><td>{{ student.email }}</td><td>{{ student.age }}</td><td>{{ student.course }}</td></tr>\n{% endfor %}\n</table>\n{% else %}<p>No students have registered yet.</p>{% endif %}\n<p><a href="/">Back to Home</a></p>\n</body></html>\n')
open('Flask_Minor_Project/requirements.txt','w').write('Flask\\n')
open('Flask_Minor_Project/README.md','w',encoding='utf-8').write('# Student Registration System (Flask Mini Project)\n\n## Description\nA beginner-friendly Flask web application that accepts student details, validates them, and displays a registration confirmation.\n\n## Features\n- Home page with student registration form\n- About page\n- Registered students page\n- POST form processing\n- Validation for blank fields, email format, and age\n- Friendly error messages\n\n## Routes\n- `/` - GET: displays the registration form\n- `/about` - GET: displays project information\n- `/students` - GET: displays students registered during this run\n- `/register` - POST: validates and processes the form\n\n## Requirements\nPython 3 and Flask.\n\n## Run locally\n1. Install dependencies: `pip install -r requirements.txt`\n2. Run: `python app.py`\n3. Open http://127.0.0.1:5000 in your browser.\n\n## Note\nThis learning project stores registrations in memory. Data is cleared when the application stops.\n')
print('Project files created.')
!find Flask_Minor_Project -maxdepth 2 -type f
```

## 4. Test the routes and form
This checks all required pages and the POST registration route before you run the live app.

```python
import os, sys
sys.path.insert(0, os.path.abspath("Flask_Minor_Project"))
from app import app as flask_app
client = flask_app.test_client()

tests = [
    ("GET /", client.get("/")),
    ("GET /about", client.get("/about")),
    ("GET /students", client.get("/students")),
    ("POST /register", client.post("/register", data={
        "name": "Sample Student",
        "email": "sample@example.com",
        "age": "20",
        "course": "Artificial Intelligence"
    }))
]
for label, response in tests:
    print(label, "->", response.status_code)
    assert response.status_code == 200
print("All route tests passed.")
```

## 6. Create the submission ZIP
The ZIP includes the complete folder structure, HTML templates, Python file, requirements and README.

```python
import os, zipfile
zip_path = "Flask_Minor_Project_Submission.zip"
with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for root, dirs, files in os.walk("Flask_Minor_Project"):
        for filename in files:
            full_path = os.path.join(root, filename)
            z.write(full_path, arcname=os.path.relpath(full_path, "."))
print("Created:", zip_path)
with zipfile.ZipFile(zip_path) as z:
    for item in z.namelist():
        print("-", item)
```
